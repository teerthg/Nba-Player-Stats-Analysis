```python
# Visualization:
# Histogram of points scored:
plt.figure(figsize=(10,6))
sns.histplot(df['PTS'],bins=20,kde=True)
plt.title("Distribution of points scored")
plt.xlabel("Points scored")
plt.ylabel("Frequency")
plt.show()
```

```python
# Check for any outliers:
Q1=df["PTS"].quantile(0.25)
Q3=df['PTS'].quantile(0.75)
IQR=Q3-Q1
print("Q1:",Q1)
print("Q3:",Q3)
print("IQR:",IQR)
Lower_Bound=Q1-1.5*IQR
Upper_Bound=Q3+1.5*IQR
print("Lower Bound:",Lower_Bound)
print("Upper Bound:",Upper_Bound)

outliers=df[(df['PTS']<Lower_Bound)|(df['PTS']>Upper_Bound)]
print("Outliers:")
print(outliers) # There is a potential outliers.
```

```python
# Correlation heatmap to check relationships between numerical variables:
plt.figure(figsize=(10,8))
numerical_df=df.select_dtypes(include=[np.number])
sns.heatmap(numerical_df.corr(),annot=True,cmap="coolwarm",fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```
